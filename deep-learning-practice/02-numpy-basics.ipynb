## Task 2
Implement a function `sum_of_squares(a)` that **vectorized** returns the sum of squares of the elements of array `a`.

```python
def sum_of_squares(a: np.ndarray) -> float:
    return float(np.sum(a ** 2))
```

## Task 3

Create a `5x5` matrix `D` where the elements on the main diagonal are equal to `1`,
and the elements on the secondary diagonal are equal to `2`.
**If an element lies at the intersection of the diagonals (the center of the matrix), its value is `3`** (1+2).

**Required:** variable `D` (type `np.ndarray`, shape = `(5,5)`).

```python
D = np.zeros((5, 5))
np.fill_diagonal(D, 1)
np.fill_diagonal(np.fliplr(D), 2)
D[2, 2] = 3
```

## Task 4

Implement a function `broadcast_op(x, y)` that takes `x` of shape `(5,1)` and `y` of shape `(1,4)`.
You need to return the matrix `M = x * y + x + y` **without loops** (using broadcasting).

**Required:** the function returns an array of shape `(5,4)`.

```python
def broadcast_op(x: np.ndarray, y: np.ndarray) -> np.ndarray:
    return x * y + x + y
```

## Task 5

Implement a function `mean_above_mean(z)` that returns the **mean value** of the elements of array `z`
which are **strictly greater** than the mean over the entire `z`.

**Hint:** use the boolean mask `z > z.mean()`.


```python
def mean_above_mean(z: np.ndarray) -> float:
    mask = z > z.mean()
    return float(z[mask].mean())
```


## Task 6

Implement a function `solve_linear(A, b)` that solves the linear system `A x = b` (without manually inverting matrices).

**Required:** use `np.linalg.solve`.

```python
def solve_linear(A: np.ndarray, b: np.ndarray) -> np.ndarray:
    return np.linalg.solve(A, b)
```

## Task 7

Create a function `row2_rev()` that forms an array of numbers from `1` to `24`,
reshapes it into a matrix of shape `(4,6)`, then **returns the second row in reverse order**.

Expected result: `[12, 11, 10, 9, 8, 7]`.

```python
def row2_rev() -> np.ndarray:
    arr = np.arange(1, 25).reshape(4, 6)
    return arr[1][::-1]
```
